```
def cambio_monedas(N, SM):
  SOLUCION = [0]*len(SM)   #SOLUCION = [0,0,0,0,..]
  ValorAcumulado = 0

  for i,valor in enumerate(SM):
    monedas =  (N-ValorAcumulado)//valor
    SOLUCION[i] = monedas
    ValorAcumulado = ValorAcumulado + monedas*valor

    if ValorAcumulado == N:
      return SOLUCION


cambio_monedas(15,[25,10,5,1])
```

## Resultado usando Fuerza Bruta

Para la obtención de los valores usando Fuerza Bruta, vamos comparando uno a uno como haría una persona normal hasta encontrar el más cercano.

```
# Listado de números
numeros = LISTA_1D

# Aquí almacenamos el conjunto de números más cercanos. Al ser en 1D, solo son números.
puntos_mas_cercanos = []

def fuerza_bruta(numeros):
  '''
    Función encargada de aplicar la recursividad
    Input: Listado de números.
    Output: Los puntos más cercanos y la distancia entre ellos.
  '''
  # Por defecto, no hay puntos cercanos hasta que se empiece la búsqueda
  puntos_mas_cercanos = None
  # Por defecto, la menor distancia es infinito.
  menor_diferencia = float("inf")
  # Recorremos las posiciones del listado para elegir número sobre el que vamos a comparar
  for i in range(len(numeros)):
    # Recorremos las posiciones del listado para comparar la posición i con la posición j
      for j in range(len(numeros)):
        # Solo podemos comparar si la posición no es la misma.
        # No comparamos el número porque al ser un listado aleatorio, se pueden repetir.
        if i != j:
          # Calculamos la diferencia.
          diferencia = abs(numeros[i] - numeros[j])
          if diferencia < menor_diferencia:
            # Actualizamos los valores a una mejor distancia.
            menor_diferencia = diferencia
            # Actualizamos los puntos más cercanos.
            puntos_mas_cercanos = []
            puntos_mas_cercanos.append([numeros[i], numeros[j]])
          elif diferencia == menor_diferencia:
            puntos_mas_cercanos.append([numeros[i], numeros[j]])

  return  puntos_mas_cercanos, menor_diferencia

par_cercano_bruta, distancia_bruta = fuerza_bruta(numeros=numeros)
print(f'Los pares más cercanos son: {par_cercano_bruta}')
print(f'Y la distancia es: {distancia_bruta}')
```

### Calcular la complejidad de la Fuerza Bruta. ¿Se puede mejorar?

La complejidad de este programa es de $O${$n^2$}. El motivo es por el uso de un doble bucle anidado.

¿Se puede mejorar? Pues la respuesta es que sí, usando otros algoritmos como Divide y Vencerás, ya que la recursividad es de menor complejidad computacional que los bucles anidados.

## Resultado usando Divide y Vencerás

```
'''
  Para realizar el algoritmo de Divide y Vencerás, hay que dividirlo en 3 partes:
  • Cruce de Frontera: Comparamos los valores entre ambos lados de la frontera
  • Divide y Vencerás: Función recursiva que se va a estar llamando hasta cumplir una de las condiciones de "return".
  • Guardamos los valores: Guardamos los valores más bajos.
'''
def cruce_frontera(lado_izq, lado_der):
    '''
    Evaluación de ambos lados para los cruces.
    Input: lado_izq y lado_der --> Listado de números.
    Output: dist y los puntos  --> La distancia entre ellos y los puntos más cercanos y l.
    '''
    dist = lado_der[0] - lado_izq[-1]
    par = [lado_izq[-1], lado_der[0]]
    return dist, [par]  # devolvemos una LISTA de pares, aunque solo haya uno


def divide_y_venceras(lista):
    n = len(lista)

    # Excepciones: Si hay menos de 2 elementos, no hay dos elementos para comparar
    if n < 2:
        return float("inf"), []
    # Si hay 2 elementos, la distancia se puede calcular.
    elif n == 2:
        return abs(lista[1] - lista[0]), [[lista[0], lista[1]]]
    # Calculamos el punto medio.
    medio = n // 2
    # Sacamos el grupo del lado izquierdo
    lado_izq = lista[:medio]
    # Calculamos el grupo del lado derecho
    lado_der = lista[medio:]

    # Recursividad de cada lado.
    dist_izq, pares_izq = divide_y_venceras(lado_izq)
    dist_der, pares_der = divide_y_venceras(lado_der)
    dist_frontera, pares_frontera = cruce_frontera(lado_izq, lado_der)

    # Buscamos la distancia mínima entre las tres candidatas
    dist_min = min(dist_izq, dist_der, dist_frontera)

    # Acumulamos TODOS los pares cuya distancia sea igual a la mínima
    pares_min = []
    # Si el par de la izquierda es igual a la distancia mínima, lo almacenamos.
    if dist_izq == dist_min:
        pares_min.extend(pares_izq)
    # Si el par de la derecha es igual a la distancia mínima, lo almacenamos.
    if dist_der == dist_min:
        pares_min.extend(pares_der)
    # Si el par de la frontera entre elementos es igual a la distancia mínima, se almacena.
    if dist_frontera == dist_min:
        pares_min.extend(pares_frontera)

    return dist_min, pares_min


def comparacion_valores(lista):
    # Ordenamos la lista
    puntos_ordenados = sorted(lista)
    # Aplicamos el algoritmo de divide y vencerás
    return divide_y_venceras(puntos_ordenados)

# Listado de los números en 1D
numeros = LISTA_1D

distancia, pares_cercanos = comparacion_valores(numeros)

print("Distancia mínima:", distancia)
print("Los pares más cercanos son:", pares_cercanos)
```

### ¿Cuál es la complejidad de Divide y Vencerás?

Para poder evaluar la complejidad, es necesario evaluar por partes.
Para el primer caso, al ordenar los valores, la complejidad es de $O(n*log(n))$. Para el segundo caso, que sería la recursión, sería de $T(n)=2T(n/2)+O(1)$ dando como resultado por recursividad y por el teorema maestro, una complejidad de $O(n)$.

Observando los valores, la complejidad de mayor grado es la de ordenación de valores siendo esta la complejidad total de este algoritmo.

## Dos dimensiones (2 Dimensiones)

### Divide y Vencerás (2D)

```
# Extensión para 2D.
# Hay que modificar y adaptar para que el cálculo de la distancia sea aprovechandonos del Teorema de Pitágoras.

def distancia(p1, p2):
  # Calculamos la distancia entre ambos puntos.
    return math.sqrt((p1[0] - p2[0])**2 + (p1[1] - p2[1])**2)

def cruce_frontera(puntos_ordenados_y, x_medio, dist_min):
    # Filtramos solo los puntos dentro de la franja de ancho 2*dist_min alrededor de x_medio
    franja = [p for p in puntos_ordenados_y if abs(p[0] - x_medio) < dist_min]

    mejor_dist = dist_min
    mejor_par = []

    # Comparamos cada punto solo con los siguientes (como mucho 7, por geometría)
    for i in range(len(franja)):
        j = i + 1
        while j < len(franja) and (franja[j][1] - franja[i][1]) < mejor_dist:
            d = distancia(franja[i], franja[j])
            if d < mejor_dist:
                mejor_dist = d
                mejor_par = [[franja[i], franja[j]]]
            elif d == mejor_dist:
                mejor_par.append([franja[i], franja[j]])
            j += 1

    return mejor_dist, mejor_par


def divide_y_venceras_2d(puntos_ordenados_x):
    '''
    Función encargada de aplicar la recursividad de Divide y Vencerás
    Input: Listado de puntos ordenados por el valor de x.
    Output: Los puntos más cercanos y la distancia entre ellos.
    '''
  # Obtenemos el N
    n = len(puntos_ordenados_x)
  # En caso de que ya N sea menor o igual a 1, devolver vacío, no hay valores para calcular.
    if n <= 1:
        return float("inf"), []
  # Cálculo de la distancia. Hay dos valroes.
    if n == 2:
        return distancia(puntos_ordenados_x[0], puntos_ordenados_x[1]), [[puntos_ordenados_x[0], puntos_ordenados_x[1]]]
  # Calculamos la posición media en valores enteros
    medio = n // 2
  # Valor de X en esa posición del medio
    x_medio = puntos_ordenados_x[medio][0]
  # Partimos por lados como en 1D
    lado_izq = puntos_ordenados_x[:medio]
    lado_der = puntos_ordenados_x[medio:]
  # Recursividad de cada lado
    dist_izq, pares_izq = divide_y_venceras_2d(lado_izq)
    dist_der, pares_der = divide_y_venceras_2d(lado_der)
  # Distancia mínima calculada
    dist_min = min(dist_izq, dist_der)
    pares_min = []
  # Si la distancia de la izquierda es igual a la mínima, se almacena.
    if dist_izq == dist_min:
        pares_min.extend(pares_izq)
  # Si la distancia de la derecha es igual a la mínima, se almacena.
    if dist_der == dist_min:
        pares_min.extend(pares_der)

    # Para la frontera, necesitamos los puntos ordenados por Y (no por X)
    puntos_ordenados_y = sorted(puntos_ordenados_x, key=lambda p: p[1])
    # Calculamos la frontera
    dist_frontera, pares_frontera = cruce_frontera(puntos_ordenados_y, x_medio, dist_min)
    # Si la distancia en la frontera es menor que la distancia mínima, devolvemos el valor como nuevo mínimo
    if dist_frontera < dist_min:
        return dist_frontera, pares_frontera
    # Si la distancia mínima es la misma que la registrada, guardamos la pareja de puntos.
    elif dist_frontera == dist_min:
        pares_min.extend(pares_frontera)
    # Devolvemos los valores mínimos y de parejas calculados.
    return dist_min, pares_min

def normalizacion(puntos_cercanos_2d):
  # Esta función se encarga de eliminar los valroes repetidos.
    vistos = set()
    resultado = []
    for p1, p2 in puntos_cercanos_2d:
        par = tuple(sorted((p1, p2)))
        if par not in vistos:
            vistos.add(par)
            resultado.append([par[0], par[1]])
    return resultado

def comparacion_valores_2d(puntos):
    # Ordenamos los puntos utilizando el eje X como comparador.
    puntos_ordenados_x = sorted(puntos, key=lambda p: p[0])
    # Aplicamos Divide y Vencerás
    dist_min_divide_2D, parejas_min_divide_2D = divide_y_venceras_2d(puntos_ordenados_x)
    # Devolvemos
    return dist_min_divide_2D, normalizacion(parejas_min_divide_2D)

# ------------------------------------------------------------------ #
# Ejecución Principal

# Listado de puntos
puntos = LISTA_2D

# Calculo de la distancia entre puntos mínima y los puntos cercanos.
distancia, puntos_cercanos = comparacion_valores_2d(puntos)

print("Distancia mínima:", distancia)
print("Los pares más cercanos son:", puntos_cercanos)
```

### Fuerza Bruta (2D)

```
def fuerza_bruta_2d(puntos):
    '''
    Función encargada de aplicar la recursividad de Fuerza Bruta
    Input: Listado de puntos.
    Output: Los puntos más cercanos y la distancia entre ellos.
    '''
    # Mejor distancia entre dos puntos, por defecto es infinito
    mejor_dist = float("inf")
    # Mejor pareja de puntos, por defecto vacío.
    mejor_par = []
    # Bucle anidado
    for i in range(len(puntos)):
        for j in range(i + 1, len(puntos)):
          # Distancia entre puntos de cada eje
            dx = puntos[i][0] - puntos[j][0]
            dy = puntos[i][1] - puntos[j][1]
          # Calculo de la distancia
            d = math.sqrt(dx**2 + dy**2)
          # Si la distancia calculada es menor, sustituir
            if d < mejor_dist:
                mejor_dist = d
                mejor_par = [[puntos[i], puntos[j]]]
          # Si es igual, añadir el nuevo par de puntos.
            elif d == mejor_dist:
                mejor_par.append([puntos[i], puntos[j]])

    return mejor_dist, mejor_par

dist, parejas = fuerza_bruta_2d(LISTA_2D)
print("Distancia mínima:", dist)
print("Pares más cercanos:", parejas)
```

## Tres Dimensiones



### Divide y Vencerás (3D)

```
def distancia_3d(p1, p2):
  # Calculo de la distancia entre puntos de 3 dimensiones.
    return math.sqrt((p1[0]-p2[0])**2 + (p1[1]-p2[1])**2 + (p1[2]-p2[2])**2)


def cruce_frontera_3d(puntos_ordenados_y, x_medio, dist_min):
  # Calculo de puntos en 3D que están dentro de los valores de la franja
    franja = [p for p in puntos_ordenados_y if abs(p[0] - x_medio) < dist_min]
  # Mejor distancia es igual a la distancia mínima.
    mejor_dist = dist_min
  # Listado de parejas
    mejor_par = []
  # Bucle que se poda en caso de obtener cuando se está lejos de la mejor distancia.
  # La complejidad computacional es de O(n)
    for i in range(len(franja)):
        j = i + 1
        while j < len(franja) and (franja[j][1] - franja[i][1]) < mejor_dist:
            # Si están muy lejos en Z con respecto a la mejor distancia, no va a dar un resultado mejor nunca.
            if abs(franja[j][2] - franja[i][2]) < mejor_dist:
                d = distancia_3d(franja[i], franja[j])
                # Si la distancia es más corta, sustituir
                if d < mejor_dist:
                    mejor_dist = d
                    mejor_par = [[franja[i], franja[j]]]
                # Si es igual, añadir el nuevo punto.
                elif d == mejor_dist:
                    mejor_par.append([franja[i], franja[j]])
            j += 1

    return mejor_dist, mejor_par


def divide_y_venceras_3d(puntos_x):
  # Obtenemos el N
    n = len(puntos_x)

    # Si no hay puntos o un punto, no se puede comparar.
    if n <= 1:
        return float("inf"), []

    # Si hay dos puntos, evaluar la distancia entre esos puntos.
    if n == 2:
        return distancia_3d(puntos_x[0], puntos_x[1]), [[puntos_x[0], puntos_x[1]]]
    # Posición del punto medio en el listado
    medio = n // 2
    x_medio = puntos_x[medio][0]
    # Obtenemos el lado izquierdo y el lado derecho
    lado_izq = puntos_x[:medio]
    lado_der = puntos_x[medio:]
    # Calculamos la distancia en ambos lados (Recursividad)
    dist_izq, pares_izq = divide_y_venceras_3d(lado_izq)
    dist_der, pares_der = divide_y_venceras_3d(lado_der)
    # Sacamos cuál tiene la distancia minima de ambos lados
    dist_min = min(dist_izq, dist_der)
    pares_min = []
    # Si la distancia de la izquierda es igual a la mínima, se almacena.
    if dist_izq == dist_min:
        pares_min.extend(pares_izq)
    # Si la distancia de la derecha es igual a la mínima, se almacena.
    if dist_der == dist_min:
        pares_min.extend(pares_der)

    # Para la frontera, ordenamos por Y (como indica el comentario de cruce_frontera_3d)
    puntos_ordenados_yz = sorted(puntos_x, key=lambda p: p[1])
    dist_frontera, pares_frontera = cruce_frontera_3d(puntos_ordenados_yz, x_medio, dist_min)
    # Si la distancia de la frontera mejora los valores, se almacena
    if dist_frontera < dist_min:
        return dist_frontera, pares_frontera
    elif dist_frontera == dist_min:
        pares_min.extend(pares_frontera)

    return dist_min, pares_min


def normaliza_pares_3d(pares):
  # Normalización para eliminar comparaciones de puntos repetidos
    vistos = set()
    resultado = []
    # Bucle de complejidad O(n)
    for p1, p2 in pares:
        par = tuple(sorted((p1, p2)))
        if par not in vistos:
            vistos.add(par)
            resultado.append([par[0], par[1]])
    return resultado


def comparacion_valores_3d(puntos):
  # Ordenamos usando el eje X como referencia
    puntos_ordenados_x = sorted(puntos, key=lambda p: p[0])
    dist_min, pares_min = divide_y_venceras_3d(puntos_ordenados_x)
    return dist_min, normaliza_pares_3d(pares_min)


numeros = LISTA_3D
dist_min_divide_3D, parejas_min_divide_3D = comparacion_valores_3d(numeros)
print("Distancia mínima:", dist_min_divide_3D)
print("Pares más cercanos:", parejas_min_divide_3D)
```

### Fuerza Bruta (3D)

```
def fuerza_bruta_3d(puntos):
  '''
    Calculo de la fuerza bruta para 3D
    Input: Listado de puntos.
    Output: Los puntos más cercanos y la distancia entre ellos.
  '''
  mejor_dist = float("inf")
  mejor_par = []
  # Bucle anidado de complejidad O(n^2)
  for i in range(len(puntos)):
      for j in range(i + 1, len(puntos)):
        # Obtenemos la distancia entre los puntos por ejes
          dx = puntos[i][0] - puntos[j][0]
          dy = puntos[i][1] - puntos[j][1]
          dz = puntos[i][2] - puntos[j][2]
          # Calculamos la distancia
          d = math.sqrt(dx**2 + dy**2 + dz**2)
          # Si la mejor distancia es mayor que el valor calculado, se almacena, si no, se descarta.
          if d < mejor_dist:
             mejor_dist = d
             mejor_par = [[puntos[i], puntos[j]]]
          # Si la distancia es igual al mejor valor, guardamos.
          elif d == mejor_dist:
             mejor_par.append([puntos[i], puntos[j]])

  return mejor_dist, mejor_par

dist_3d, parejas_3d = fuerza_bruta_3d(LISTA_3D)
print("Distancia mínima:", dist_3d)
print("Pares más cercanos:", parejas_3d)
```
